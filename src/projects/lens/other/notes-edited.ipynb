# Задача

Найти профиль осесимметричной линзы $z(r)$, то есть образованной поверхностями

$$
z_1(r) = z(r), \quad z_2(r) = t-z(r)
$$

из материала с показателем преломления $n$, такой, чтобы параллельный пучок любого диаметра, падающий вдоль оси $z$, сходился в одну точку на оси $z$ на расстоянии $f$ от центра линзы.


## Замечание про толщину линзы $t$

В данной задаче $t$ оказывается параметром, который мы можем выбирать произвольно (среди допустимых ему значений). То есть различные $t$ будут задавать различные поверхности $z(r)$.

Из геометрических соображений, уже можно получить некоторое ограничение на t.

$$t > 0, \quad f - \frac{t}{2}>0$$

Второе условие означает, что фокус находится за второй поверхностью линзы, а не внутри нее или перед ней.

Таким образом, получаем:

$$0 < t < 2f$$

# Решение

## 1. Составление уравнения

![Рисунок](image.png)

Напишем условие принадлежности точки излома $(r_b, z_b)$ второй поверхности линзы:

$$z_b = z_2(r_b).$$

Используем $z_b = z(r) + w$ (см. рис. 1) и тот факт, что $(r_b, z_b)$ так же принадлежит и участку луча, находящемуся внутри линзы. То есть, что $r_b$ можно записать в виде:

$$r_b = r + qw,$$

Где $q$ — наклон участка луча внутри линзы. Тогда это условие принадлежности можно записать в виде:

$$
\begin{equation}
z(r) + w = t - z(r + qw).
\tag{1}
\end{equation}
$$

Уравнение (1) задает поверхность $z(r)$. Осталось выразить через известные параметры наклон луча $q$ внутри линзы и горизонтальное смещение луча $w$ внутри линзы соответсвтенно.

## 2. Поиск внутреннего наклона луча $q$

Пусть $\mathbf{d}_i$, $\mathbf{d}_t$ — единичные векторы падающего и первого преломленного участков лучей соответственно. Разложим $\mathbf{d}_t$ по базису $\mathbf{e}_r$, $\mathbf{e}_z$: 

$$\mathbf{d}_t = d_r \mathbf{e}_r + d_z \mathbf{e}_z.$$

Где $\mathbf{e}_r = (\cos \varphi, \sin\varphi, 0),\, \mathbf{e}_z = (0, 0, 1)$ (причём получается $\mathbf{e}_z$ = $\mathbf{d}_i$)

Тогда наклон $q$ равен:

$$q = \frac{d_r}{d_z}.$$

Используем закон Снеллиуса в векторной формуле $\left(\mathbf{d}^{\tau}_i = n \mathbf{d}^{\tau}_t\right)$:

$$
\begin{equation}
\mathbf{d}_t-(\mathbf{d}_t \mathbf{n}) \mathbf{n} = \frac{1}{n} \left(\mathbf{d}_i - (\mathbf{d}_i \mathbf{n}) \mathbf{n} \right).
\tag{2}
\end{equation}
$$

Обозначим $p = z'(r)$, тогда:

$$\mathbf{n} = \frac{-p \mathbf{e}_r + \mathbf{e}_z}{\sqrt{1 + p^2}}, \quad \mathbf{d}_i \mathbf{n} = \frac{1}{\sqrt{1 + p^2}}.$$

Обозначим так же $c_t = \mathbf{d}_t \mathbf{n}$. Из соотношения (2) получаем:

$$
\begin{equation}
\mathbf{d}_t = \frac{1}{n} \left(\mathbf{d}_i - (\mathbf{d}_i \mathbf{n}) \mathbf{n} \right)+ c_t \mathbf{n}.
\tag{3}
\end{equation}
$$

Так как $\mathbf{d}_i - (\mathbf{d}_i \mathbf{n}) \mathbf{n} = \mathbf{d}^{\tau}_i \, \bot \,\mathbf{n}$, то возводя в квадрат (3), получаем для $c_t$:

$$
\begin{aligned}
c^2_t &= 1 - \frac{1}{n^2} (\mathbf{d}_i - (\mathbf{d}_i \mathbf{n}) \mathbf{n})^2= \\
&= 1 - \frac{1}{n^2}\left(1 - 2(\mathbf{d}_i \mathbf{n})^2 + (\mathbf{d}_i \mathbf{n})^2\right)= \\
&= 1 - \frac{1}{n^2}\left(1 - (\mathbf{d}_i \mathbf{n})^2\right)= \\
&= 1 - \frac{1}{n^2} \left( 1 - \frac{1}{1 + p^2} \right)= \\
&= 1 - \frac{p^2}{n^2 (1 + p^2)}= \\
&= \frac{n^2 + n^2 p^2 - p^2}{n^2 (1 + p^2)}= \\
&= \frac{n^2 + (n^2 - 1)p^2}{n^2 (1 + p^2)}.
\end{aligned}
$$

Обозначим $A = \sqrt{n^2 + (n^2 - 1)p^2}$, тогда получаем для $\mathbf{d}_t$:

$$
\begin{aligned}
\mathbf{d}_t &= \frac{1}{n}\left( \mathbf{e}_z - \frac{1}{\sqrt{1 + p^2}} \mathbf{n} \right ) + \frac{A}{n\sqrt{1 + p^2}}\mathbf{n}= \\
&= \frac{1}{n} \mathbf{e}_z + \frac{A-1}{n\sqrt{1+p^2}}\mathbf{n}= \\
&= \frac{1}{n} \mathbf{e}_z + \frac{A-1}{n\sqrt{1+p^2}}\left(\frac{-p \mathbf{e}_r + \mathbf{e}_z}{\sqrt{1 + p^2}}\right)= \\
&= \frac{p(1-A)}{n(1+p^2)}\mathbf{e}_r + \left[\frac{A-1}{n(1+p^2)}+ \frac{1}{n}\right]\mathbf{e}_z = \\
&= \frac{p(1-A)}{n(1+p^2)}\mathbf{e}_r + \frac{A+p^2}{n(1+p^2)}\mathbf{e}_z.
\end{aligned}
$$

То есть

$$q = \frac{p(1-A)}{A+p^2}.$$


## 3. Поиск горизонтального смещения луча $w$

Рассмотрим оптический путь луча $S(r)$, начинающегося от плоскости $z = 0$ и заканчивающегося на плоскости $z = z_f = f + \frac{t}{2}$ (см. рис. 1):

$$
\begin{equation}
\begin{aligned}
S(r) &= z(r) + n \sqrt{w^2 + (qw)^2} + \sqrt{r_b^2+(z_f-z_b)^2} = \\
&= z(r) + nw \sqrt{1 + q^2} + \sqrt{(r+qw)^2+(z_f-z(r)-w)^2}.
\end{aligned}
\tag{4}
\end{equation}
$$

Оптический путь при $r=0$:

$$ S(0) = nt + (z_f - t) = z_f + (n - 1)t = f + \left(n - \frac{1}{2}\right)t. $$

Пользуясь принципом Ферма (для любого $r$ выполняется $S(0) = S(r)$) и возводя (4) в квадрат, получаем:

$$ (r+qw)^2 + (z_f - z(r)-w)^2 = S^2(0) + z^2(r) + n^2 (1+q^2)w^2 - 2S(0)z(r) - 2S(0)n\sqrt{1+q^2}w + 2z(r)n\sqrt{1+q^2}w. $$

Получили квадратное уравнение на $w$. Приведём подобные.

$$ r^2 + 2rqw + q^2w^2 + z^2_f + z^2(r) + w^2 - 2z_f z(r) - 2z_f w + 2z(r)w = \\
= S^2(0) + z^2(r) + n^2 (1 + q^2)w^2 - 2S(0)z(r) - 2S(0)n\sqrt{1+q^2}w + 2z(r)n\sqrt{1+q^2}w $$

$$ \left(1+q^2\right)\left(1-n^2\right)w^2 + 2\left(rq-\left(z_f - z(r)\right) + \left(S(0) - z(r)\right)n\sqrt{1+q^2}\right)w + \left( r^2 + (z_f - z(r))^2 - (S(0)-z(r))^2\right)=0$$

В итоге:

$$a_w w^2 + b_w w + c_w = 0, $$

где

$$
\begin{aligned}
a_w &= \left(1+q^2\right)\left(1-n^2\right), \\
b_w &= 2\left(rq-\left(z_f - z(r)\right) + \left(S(0) - z(r)\right)n\sqrt{1+q^2}\right), \\
c_w &= \left( r^2 + (z_f - z(r))^2 - (S(0)-z(r))^2\right).
\end{aligned}
$$

При этом $w$ должен удовлетворять условиям исходного, ещё не возведённого в квадрат уравнения.


## 4. Итоговое уравнение

Таким образом, мы получили конечное уравнение, которое задает поверхность $z(r)$.

$$
\boxed{z(r) + w = t - z(r + qw)}
$$

Где

$$q = \frac{p(1-A)}{A+p^2}, $$

a $w$ задаётся квадратным уравнением

$$a_w w^2 + b_w w + c_w = 0, $$

где

$$
\begin{aligned}
a_w &= \left(1+q^2\right)\left(1-n^2\right), \\
b_w &= 2\left(rq-\left(f + \frac{t}{2} - z(r)\right) + \left(S(0) - z(r)\right)n\sqrt{1+q^2}\right), \\
c_w &= \left( r^2 + \left(f + \frac{t}{2} - z(r)\right)^2 - (S(0)-z(r))^2\right).
\end{aligned}
$$

# Численное решение

## 1. Вычисление начального профиля поверхности

Для численного решения необходимо построить начальный профиль поверхности в окрестности оптической оси. В силу осевой симметрии гладкий профиль представим рядом Тейлора

$$
z(r)=\sum_{k=1}^{\infty}a_{2k}r^{2k}.
$$

Коэффициенты $a_{2k}$ определяются из замыкающего уравнения

$$
R(r)=z(r) + w - t + z(r + qw)=0.
$$

Подставляя ряд в $R(r)$ и приравнивая к нулю коэффициенты при одинаковых степенях $r$, получаем последовательность уравнений для $a_{2k}$. В силу чётности $R(r)$ её разложение содержит только чётные степени:

$$
R(r)=R_0+\sum_{k=1}^{\infty}R_{2k}r^{2k},
$$

причём $R_0=0$ выполняется автоматически, а содержательные уравнения имеют вид

$$
R_2=R_4=\dots=0.
$$

Коэффициент $a_2$ находится из нелинейного уравнения $R_2=0$ методом бисекции. Интервал локализуется автоматически расширением вокруг параксиальной оценки $a_2 \approx 1/\bigl(4(n-1)f\bigr)$: полуширина удваивается, пока не появится смена знака, нижняя граница не допускается отрицательной; затем бисекция доводит интервал до заданного допуска. Каждый следующий коэффициент $a_{2k}$ находится из линейного по $a_{2k}$ уравнения $R_{2k}=0$ при фиксированных младших коэффициентах (в коде — секущим шагом по двум значениям невязки).

Ниже приведён код, воспроизводящий коэффициенты до $a_{18}$, печатающий их, максимальную невязку ряда и начальные значения профиля и его производной в точке ${r_0}^{(*)}$ вблизи оси. Вычисления ведутся в `mpmath` с повышенной точностью.

$(*)$ — Выбор $r_0$ — эвристический: он должен быть достаточно мал, чтобы усечённый ряд был точен, и достаточно велик, чтобы численное интегрирование было устойчивым.

```python
# ============================================================
# Ячейка 1. Параметры линзы и численных методов
# ------------------------------------------------------------
# Единственный источник истины. Значения хранятся строками,
# чтобы каждая последующая ячейка сама переводила их в нужный
# ей тип (mp.mpf или float).
# ============================================================

# --- Параметры линзы -----------------------------------------
N_STR = "1.5"      # коэффициент преломления линзы
T_STR = "1"      # толщина линзы
F_STR = "10.0"     # фокусное расстояние

# --- Параметры численного метода (расчёт начальной поверхности)
KMAX = 5                          # число коэффициентов a_2..a_{2*KMAX}
SQRT_ITERATIONS = 6               # итераций Ньютона для корня ряда
BISECTION_MAX_ITERATIONS = 300    # предел итераций бисекции
BISECTION_TOL_STR = "1e-75"       # допуск по ширине интервала
A2_INITIAL_DELTA_STR = "0.01"     # начальная полуширина интервала для a_2
A2_BRACKET_MAX_EXPANSIONS = 100   # предел удвоений интервала
R0_STR = "0.05"                   # точка старта задачи Коши

# --- Точность mpmath (для ячейки расчёта начальной поверхности)
MP_DPS = 80

# --- Параметры независимой проверки (numpy/scipy) ------------
# (здесь только то, что не выводится из параметров линзы;
#  остальное — апертура, число лучей и т.п. — задаётся
#  прямо в ячейке проверки)
```

```python
# ============================================================
# Ячейка 2. Расчёт начальной поверхности
# ------------------------------------------------------------
# Зависит только от констант ячейки 1.
# Внутри переводит их в mp.mpf (высокая точность).
# На выходе: coeffs_init_mp (list[mp.mpf]) и coeffs_init_np (np.ndarray[float64]).
# ============================================================
import mpmath as mp
import numpy as np

# --- Точность ---
mp.mp.dps = MP_DPS

# --- Параметры линзы (из ячейки 1, перевод в mp.mpf) ---
n = mp.mpf(N_STR)
t = mp.mpf(T_STR)
f = mp.mpf(F_STR)

# --- Параметры численного метода (из ячейки 1) ---
SERIES_LENGTH              = 2 * KMAX + 1
BISECTION_TOL              = mp.mpf(BISECTION_TOL_STR)
A2_INITIAL_DELTA           = mp.mpf(A2_INITIAL_DELTA_STR)
R0                         = mp.mpf(R0_STR)
# SQRT_ITERATIONS, BISECTION_MAX_ITERATIONS, A2_BRACKET_MAX_EXPANSIONS,
# KMAX — целые, берутся напрямую.

# --- Вспомогательные величины ---
tb = f - t / 2          # расстояние от задней поверхности до фокальной плоскости
zf = t + tb             # координата фокальной плоскости
S0 = n * t + tb         # оптическая длина пути для центрального луча


# ---------------------------------------------------------
# Арифметика усечённых степенных рядов
# ---------------------------------------------------------
def const(v, L):
    return [mp.mpf(v)] + [mp.mpf(0)] * (L - 1)

def zero(L):
    return [mp.mpf(0)] * L

def scale(a, c):
    return [x * c for x in a]

def add(a, b):
    return [x + y for x, y in zip(a, b)]

def neg(a):
    return [-x for x in a]

def mul(a, b):
    L = len(a)
    out = zero(L)
    for i in range(L):
        if a[i] == 0:
            continue
        for j in range(L - i):
            out[i + j] += a[i] * b[j]
    return out

def reciprocal(a):
    L = len(a)
    out = zero(L)
    out[0] = 1 / a[0]
    for k in range(1, L):
        out[k] = -sum(a[i] * out[k - i] for i in range(1, k + 1)) / a[0]
    return out

def div(a, b):
    return mul(a, reciprocal(b))

def sqrt_series(a):
    L = len(a)
    s = [mp.sqrt(a[0])] + [mp.mpf(0)] * (L - 1)
    for _ in range(SQRT_ITERATIONS):
        s = scale(add(s, div(a, s)), mp.mpf('0.5'))
    return s

def deriv(a):
    L = len(a)
    return [mp.mpf(k) * a[k] for k in range(1, L)] + [mp.mpf(0)]

def compose(a, b):
    L = len(a)
    out = zero(L)
    for i in range(L - 1, -1, -1):
        out = add(mul(out, b), const(a[i], L))
    return out

def r_series(L):
    out = zero(L)
    out[1] = mp.mpf(1)
    return out


# ---------------------------------------------------------
# Замыкающая невязка R(r)
# ---------------------------------------------------------
def build_R(zs):
    """Коэффициенты разложения R(r) по степеням r."""
    L = len(zs)
    ps = deriv(zs)
    p2 = mul(ps, ps)

    A = sqrt_series(add(const(n * n, L), scale(p2, n * n - 1)))

    q = div(
        scale(mul(ps, add(A, const(-1, L))), -1),
        add(A, p2)
    )

    q2 = mul(q, q)
    A1 = scale(sqrt_series(add(const(1, L), q2)), n)
    D  = add(const(S0, L), neg(zs))

    aw = scale(add(const(1, L), q2), 1 - n * n)

    rs   = r_series(L)
    rq   = mul(rs, q)
    zf_z = add(const(zf, L), neg(zs))

    bw = scale(add(add(rq, neg(zf_z)), mul(D, A1)), 2)

    cw = add(
        add(mul(rs, rs), mul(zf_z, zf_z)),
        neg(mul(D, D))
    )

    discriminant = add(
        mul(bw, bw),
        scale(mul(aw, cw), -4)
    )

    w = div(
        add(neg(bw), sqrt_series(discriminant)),
        scale(aw, 2)
    )

    rb = add(rs, mul(q, w))

    return add(
        add(add(zs, w), const(-t, L)),
        compose(zs, rb)
    )

def series_from_coeffs(coeffs):
    """Собрать ряд z(r) из чётных коэффициентов coeffs[k] = a_{2k}."""
    L = 2 * len(coeffs) + 1
    zs = zero(L)
    for k, a in enumerate(coeffs, start=1):
        zs[2 * k] = a
    return zs

def R2_at(a2):
    """Коэффициент при r^2 в невязке R(r) при z(r) = a2 * r^2."""
    L = 2 * KMAX + 1
    zs = zero(L)
    zs[2] = a2
    return build_R(zs)[2]


# ---------------------------------------------------------
# Вычисление коэффициентов
# ---------------------------------------------------------
def solve_recurrence(KMAX):
    """Возвращает coeffs (a_2..a_{2*KMAX}), residual и диагностику бисекции."""
    L = 2 * KMAX + 1
    coeffs = [mp.mpf(0)] * (KMAX + 1)   # coeffs[k] = a_{2k}

    # --- a_2: бисекция по R_2(a_2) = 0 ---
    a2_approx = 1 / (4 * (n - 1) * f)

    delta = A2_INITIAL_DELTA
    bracket_found = False
    lo = hi = None

    for _ in range(A2_BRACKET_MAX_EXPANSIONS + 1):
        lo = max(a2_approx - delta, mp.mpf('0'))
        hi = a2_approx + delta
        R_lo, R_hi = R2_at(lo), R2_at(hi)

        if R_lo == 0:
            hi = lo; bracket_found = True; break
        if R_hi == 0:
            lo = hi; bracket_found = True; break
        if R_lo * R_hi < 0:
            bracket_found = True; break

        delta *= 2

    if not bracket_found:
        raise RuntimeError(
            "Не удалось найти интервал со сменой знака R2(a2). "
            "Проверьте параксиальную оценку и поведение R2_at."
        )

    if lo != hi:
        R_lo, R_hi = R2_at(lo), R2_at(hi)
        for _ in range(BISECTION_MAX_ITERATIONS):
            mid = (lo + hi) / 2
            R_mid = R2_at(mid)
            if R_mid == 0:
                lo = hi = mid; break
            if R_lo * R_mid < 0:
                hi, R_hi = mid, R_mid
            else:
                lo, R_lo = mid, R_mid
            if hi - lo < BISECTION_TOL:
                break
        else:
            raise RuntimeError(
                "Бисекция не достигла заданной точности "
                "за BISECTION_MAX_ITERATIONS."
            )

    coeffs[1] = (lo + hi) / 2

    # --- остальные a_{2k}: линейные уравнения ---
    for k in range(2, KMAX + 1):
        idx = 2 * k

        z0 = zero(L)
        z1 = zero(L)
        for j in range(1, k):
            z0[2 * j] = coeffs[j]
            z1[2 * j] = coeffs[j]
        z1[idx] = mp.mpf(1)          # пробная единица при a_{2k}

        R0 = build_R(z0)[idx]
        R1 = build_R(z1)[idx]
        coeffs[k] = -R0 / (R1 - R0)

    # --- контроль качества ---
    R_full = build_R(series_from_coeffs(coeffs[1:]))
    residual = max(abs(c) for c in R_full)

    info = {
        "a2_approx": a2_approx,
        "bracket": (lo, hi),
        "R_bracket": (R2_at(lo), R2_at(hi)),
        "residual": residual,
        "R_full": R_full,
    }
    return coeffs, info


# ---------------------------------------------------------
# Начальный профиль и производная
# ---------------------------------------------------------
def z_init(r, coeffs):
    r = mp.mpf(r)
    return sum(coeffs[k] * r ** (2 * k) for k in range(1, len(coeffs)))

def p_init(r, coeffs):
    r = mp.mpf(r)
    return sum(2 * k * coeffs[k] * r ** (2 * k - 1)
               for k in range(1, len(coeffs)))


# ---------------------------------------------------------
# Запуск
# ---------------------------------------------------------
coeffs_init, info = solve_recurrence(KMAX)
coeffs_init_mp = coeffs_init                                   # mp.mpf, как есть
coeffs_init_np = np.array([float(c) for c in coeffs_init])     # float64 для проверки

# --- Печать: коэффициенты ---
print("\nКоэффициенты разложения:")
for k in range(1, KMAX + 1):
    print(f"  a_{2*k} = {mp.nstr(coeffs_init[k], 18)}")

terms = []
for k in range(1, KMAX + 1):
    val = coeffs_init[k]
    s = mp.nstr(val, 9, strip_zeros=False)
    if not s.startswith("-"):
        s = "+" + s
    terms.append(f"{s}*r^{2*k}")

body = "".join(terms)
if body.startswith("+"):
    body = body[1:]
print("\n z0+" + body)

# --- Печать: невязка ---
R_full = info["R_full"]
idx_max = max(range(len(R_full)), key=lambda i: abs(R_full[i]))
print("\nНевязка R(r):")
print(f"  max |R_i| = {mp.nstr(info['residual'], 5)}")

# --- Печать: начальные данные ---
z0 = z_init(R0, coeffs_init)
p0 = p_init(R0, coeffs_init)
print("\nНачальные данные для продолжения:")
print(f"  r0 = {mp.nstr(R0, 12)}")
print(f"  z0 = z(r0)  = {mp.nstr(z0, 18)}")
print(f"  p0 = z'(r0) = {mp.nstr(p0, 18)}")
```

## 2. Проверка построенной начальной поверхности

Для проверки точности начального профиля независимо трассируем лучи через обе поверхности линзы. Направления лучей на границах раздела определяются векторным законом Снеллиуса, а точка пересечения луча со второй поверхностью находится численно. После выхода из линзы вычисляем координату пересечения луча с оптической осью $z_F(r)$ и полную оптическую длину пути

$$
S(r)=z(r)+nL_{\mathrm{int}}+L_{\mathrm{exit}},
$$

где $L_{\mathrm{int}}$ и $L_{\mathrm{exit}}$ — геометрические длины пути внутри линзы и после выхода из неё до плоскости $z=z_F(r)$.

Для идеальной линзы все лучи должны пересекать ось в точке $z_f$, а оптическая длина пути должна быть постоянной и равной $S(0)$. Поэтому оцениваем максимальные отклонения

$$
\varepsilon_f=\max_{r}|z_F(r)-z_f|,
\qquad
\varepsilon_{S}=\max_{r}|S(r)-S(0)|.
$$

Эта проверка использует непосредственно профиль и закон Снеллиуса, не используя замыкающее уравнение $R(r)=0$.

```python
# ============================================================
# Ячейка 3. Независимая проверка трассировки (с поддержкой
#           как коэффициентов, так и готовых функций)
# ------------------------------------------------------------
# Зависит от констант ячейки 1.
#
# Два способа вызова:
#   1) verify_ray_tracing(coeffs)               — по ряду
#   2) verify_ray_tracing(z_func=..., p_func=...) — по функциям
#
# Параметр show_plot=True/False включает/выключает график.
#
# ПВО (полное внутреннее отражение) на второй поверхности:
#   - refract возвращает None вместо ValueError;
#   - trace_full возвращает (None, r2, S) при ПВО;
#   - verify_ray_tracing пропускает такие лучи при расчёте
#     max|z_F|, max|S| и при отрисовке;
#   - печатает число лучей с ПВО (если > 0).
# ============================================================
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import brentq

# --- Параметры линзы (из ячейки 1, перевод в float) ---
n_f  = float(N_STR)
t_f  = float(T_STR)
f_f  = float(F_STR)

# --- Производные величины (float) ---
tb_f = f_f - t_f / 2
zf_f = t_f + tb_f
S0_f = n_f * t_f + tb_f


# ---------------------------------------------------------
# Float-версии профиля и производной (по коэффициентам ряда)
# ---------------------------------------------------------
def z_init_np(r, coefficients):
    """z(r) = sum_{k>=1} coefficients[k] * r^{2k}, float-версия."""
    r = np.asarray(r, dtype=float)
    z = np.zeros_like(r)
    r2 = r * r
    rp = np.ones_like(r)
    for k in range(1, len(coefficients)):
        rp = rp * r2
        z = z + coefficients[k] * rp
    return z


def p_init_np(r, coefficients):
    """z'(r) = sum_{k>=1} 2k * coefficients[k] * r^{2k-1}, float-версия."""
    r = np.asarray(r, dtype=float)
    p = np.zeros_like(r)
    r2 = r * r
    rp = r.copy()
    for k in range(1, len(coefficients)):
        p = p + (2 * k) * coefficients[k] * rp
        rp = rp * r2
    return p


# ---------------------------------------------------------
# Преломление
# ---------------------------------------------------------
def refract(d, normal, eta):
    """Преломлённое направление по закону Снеллиуса. eta = n1/n2.

    Возвращает направление или None при полном внутреннем отражении.
    """
    cos_i = -np.dot(normal, d)
    k = 1.0 - eta**2 * (1.0 - cos_i**2)
    if k < 0:
        return None
    cos_t = np.sqrt(k)
    out = eta * d + (eta * cos_i - cos_t) * normal
    return out / np.linalg.norm(out)


# ---------------------------------------------------------
# Трассировка (внутренние функции — с явными z_func/p_func)
# ---------------------------------------------------------
def trace_full(r1, z_func, p_func):
    """Трассировка луча для r1 >= 0.

    Возвращает:
      - (z_focus, r2, S)     — нормальный луч;
      - (None, r2, S)        — ПВО на второй поверхности;
      - (None, None, None)   — ПВО на первой поверхности (не бывает
                               при входе из воздуха в линзу, но
                               защищаемся).
    """
    z1  = float(z_func(r1))
    dz1 = float(p_func(r1))

    # Преломление на первой поверхности: воздух -> линза.
    normal1 = np.array([dz1, -1.0])
    normal1 /= np.linalg.norm(normal1)
    d1 = refract(np.array([0.0, 1.0]), normal1, 1.0 / n_f)
    if d1 is None:
        # ПВО на первой поверхности (физически при n>1 невозможно,
        # но защищаемся).
        return None, None, None

    # Пересечение внутреннего луча со второй поверхностью.
    if abs(d1[0]) < 1e-15:
        r2 = 0.0
    else:
        dr_dz = d1[0] / d1[1]

        def intersection(r2):
            return z1 + (r2 - r1) / dr_dz - (
                t_f - float(z_func(r2))
            )

        r2 = brentq(intersection, 0.0, r1, xtol=1e-14, rtol=1e-14)

    z2 = t_f - float(z_func(r2))

    L_int  = np.hypot(r2 - r1, z2 - z1)
    L_exit = np.hypot(r2, zf_f - z2)
    S = z1 + n_f * L_int + L_exit

    # Преломление на второй поверхности: линза -> воздух.
    dz2 = -float(p_func(r2))
    normal2 = np.array([dz2, -1.0])
    normal2 /= np.linalg.norm(normal2)
    d2 = refract(d1, normal2, n_f)
    if d2 is None:
        # ПВО — луч не вышел из линзы.
        return None, r2, S

    if abs(d2[0]) < 1e-14:
        z_focus = zf_f
    else:
        z_focus = z2 - d2[1] * r2 / d2[0]

    return z_focus, r2, S


def trace_signed(r1, z_func, p_func):
    """Трассировка с учётом знака r (профиль симметричен)."""
    sign = 1.0 if r1 >= 0 else -1.0
    z_focus, r2, S = trace_full(abs(r1), z_func, p_func)
    if z_focus is None:
        return None, (sign * r2 if r2 is not None else None), S
    return z_focus, sign * r2, S


def draw_ray(r1, z_func, p_func, color="red", linewidth=0.8, z_start=0.0):
    """Рисует луч: до линзы, внутри и до фокуса.

    Если луч испытал ПВО — ничего не рисует.
    """
    z_focus, r2, _ = trace_signed(r1, z_func, p_func)
    if z_focus is None:
        return   # ПВО — не рисуем

    z1 = float(z_func(abs(r1)))
    z2 = t_f - float(z_func(abs(r2)))

    plt.plot(
        [z_start - t_f / 2, z1 - t_f / 2],
        [r1, r1], color=color, linewidth=linewidth
    )
    plt.plot(
        [z1 - t_f / 2, z2 - t_f / 2],
        [r1, r2], color=color, linewidth=linewidth
    )
    plt.plot(
        [z2 - t_f / 2, z_focus - t_f / 2],
        [r2, 0], color=color, linewidth=linewidth
    )


# ---------------------------------------------------------
# Независимая проверка
# ---------------------------------------------------------
def verify_ray_tracing(coefficients=None, z_func=None, p_func=None,
                       title=None, show_plot=True):
    """Независимая проверка трассировки лучей.

    Параметры
    ----------
    coefficients : array-like or None
        Коэффициенты ряда z(r) = Σ a_{2k} r^{2k}.
        Используется, если z_func / p_func не заданы.
    z_func, p_func : callable or None
        Готовые функции поверхности и производной.
        Если заданы — coefficients игнорируется.
    title : str or None
        Заголовок графика (если show_plot=True).
    show_plot : bool
        True  — рисует график и вызывает plt.show();
        False — не рисует ничего, только считает метрики.
        Возвращаемый dict в любом случае содержит поля 'fig', 'ax',
        но при show_plot=False они равны None.

    Возвращает
    ----------
    dict с ключами:
        'aperture', 'zfs', 'Ss', 'max_dz', 'max_dS',
        'n_pvo', 'n_total', 'fig', 'ax'
    """
    # --- Выбор источника z(r) и p(r) ---
    if z_func is None or p_func is None:
        if coefficients is None:
            raise ValueError(
                "Нужно задать либо coefficients, либо z_func и p_func."
            )
        _coefficients = coefficients
        def z_func(r, _c=_coefficients):
            return z_init_np(r, _c)
        def p_func(r, _c=_coefficients):
            return p_init_np(r, _c)

    # 1. Поиск апертуры: первая положительная точка, где z(r) = t/2.
    r_scan = np.linspace(1e-6, 10.0, 2000)
    g_scan = np.array([float(z_func(r)) - t_f / 2 for r in r_scan])
    crossings = np.flatnonzero(g_scan[:-1] * g_scan[1:] <= 0)

    if len(crossings) == 0:
        raise ValueError(
            "Не удалось найти апертуру: проверь профиль и диапазон поиска."
        )

    i = crossings[0]
    ap = brentq(
        lambda r: float(z_func(r)) - t_f / 2,
        r_scan[i], r_scan[i + 1],
        xtol=1e-14, rtol=1e-14,
    )

    # 2. Трассировка по всей апертуре (оба знака r).
    check_r = np.linspace(-ap * 0.995, ap * 0.995, 600)
    check_r = check_r[np.abs(check_r) > ap / 1000]
    traces = [trace_signed(r, z_func, p_func) for r in check_r]

    n_total = len(traces)
    n_pvo = sum(1 for tr in traces if tr[0] is None)

    # Отбираем только успешные лучи для метрик
    zfs = np.array([tr[0] for tr in traces if tr[0] is not None])
    Ss  = np.array([tr[2] for tr in traces if tr[0] is not None])

    if len(zfs) == 0:
        raise RuntimeError(
            f"Все {n_total} лучей испытали ПВО — линза не работает."
        )

    max_dz = float(np.max(np.abs(zfs - zf_f)))
    max_dS = float(np.max(np.abs(Ss - S0_f)))

    # 3. Печать метрик
    print(f"Апертура a = {ap:.10f}")
    print(f"Лучей: {n_total},  без ПВО: {n_total - n_pvo},  с ПВО: {n_pvo}")
    if n_pvo > 0:
        print(f"  (лучи с ПВО исключены из метрик и не рисуются)")
    print(f"max|z_F(r) - z_f| = {max_dz:.6e}   (по {len(zfs)} лучам)")
    print(f"max|S(r) - S(0)| = {max_dS:.6e}")

    # 4. График (если нужен)
    fig = None
    ax = None
    if show_plot:
        # Профиль линзы.
        rr = np.linspace(0.0, ap, 400)
        zr = np.array([float(z_func(r)) for r in rr])

        fig, ax = plt.subplots(figsize=(7, 6))

        ax.plot(zr - t_f / 2, rr, color="blue")
        ax.plot(t_f - zr - t_f / 2, rr, color="blue")
        ax.plot(zr - t_f / 2, -rr, color="blue")
        ax.plot(t_f - zr - t_f / 2, -rr, color="blue")

        # Лучи по всей апертуре (оба знака r).
        # Лучи с ПВО не рисуются (draw_ray сам пропускает).
        for r1 in np.linspace(-ap * 0.995, ap * 0.995, 31):
            if abs(r1) < ap / 1000:
                continue
            draw_ray(r1, z_func, p_func, color="red",
                     linewidth=0.8, z_start=0.0)

        # Центральный луч.
        draw_ray(0.0, z_func, p_func, color="red",
                 linewidth=1.2, z_start=0.0)

        ax.set_xlabel(r"$z-t/2$")
        ax.set_ylabel(r"$r$")
        ax.axis("equal")
        ax.grid(True)
        if title:
            ax.set_title(title)

        plt.show()

    return {
        "aperture": ap,
        "zfs": zfs,
        "Ss": Ss,
        "max_dz": max_dz,
        "max_dS": max_dS,
        "n_pvo": n_pvo,
        "n_total": n_total,
        "fig": fig,
        "ax": ax,
    }
```

## 3. Численное продолжение профиля

Начальный профиль $z(r)$ и его производная $p(r)=z'(r)$ при $r\le r_0$ задаются рядом из пункта 1. При $r>r_0$ решение продолжается численно. На каждом шаге совместно определяются $z_{k+1}$ и $p_{k+1}$ из замыкающего уравнения и трапецеидальной формулы

$$
\begin{cases}
z + w - t + z(r_b) = 0,\\[2pt]
z - z_k - \dfrac{h}{2}\,(p_k + p) = 0,
\end{cases}
\qquad r_b = r + q(p)\,w.
$$

Значение $z(r_b)$ вычисляется по уже построенному профилю; если $r_b$ попадает внутрь текущего шага, используется кубическая интерполяция Эрмита с учётом неизвестных значений на его конце. Система решается методом Ньютона с численным якобианом и уменьшением шага при неудаче.

После достижения $z(a)=t/2$ продолжённый профиль аппроксимируется чётным полиномом. Коэффициенты находятся методом наименьших квадратов при точных условиях сшивки с начальным рядом по значению и первой производной. Степень полинома выбирается по максимальному отклонению от численного решения на плотной сетке; если заданная точность не достигнута, это явно сообщается.

Предполагается, что `coeffs` и `R0` уже определены в пункте 1, причём `coeffs[k] = a_{2k}`.

## 4. Независимая проверка численного решения

Для проверки точности профиля, продолженного численно в пункте 3, повторяем независимую трассировку лучей через обе поверхности линзы. Используются те же функции `refract`, `trace_full`, `trace_signed`, `draw_ray`, что и в пункте 2, но вместо ряда из пункта 1 подставляются функции `z_sol` и `dz_sol`, полученные численным продолжением (и, при необходимости, полиномиальной аппроксимацией). Направления лучей на границах раздела определяются векторным законом Снеллиуса, точка пересечения луча со второй поверхностью находится численно, а после выхода из линзы вычисляется координата пересечения с оптической осью $z_F(r)$ и полная оптическая длина пути

$$
S(r)=z(r)+nL_{\mathrm{int}}+L_{\mathrm{exit}}.
$$

Как и в пункте 2, оцениваем максимальные отклонения

$$
\varepsilon_f=\max_{r}|z_F(r)-z_f|,
\qquad
\varepsilon_{S}=\max_{r}|S(r)-S(0)|.
$$

Здесь $z_f=\mathrm{zF}$ — координата заданной фокальной плоскости, а $S(0)=\mathrm{OPL0}$ — оптическая длина пути центрального луча. Проверка не использует замыкающее уравнение $R(r)=0$: она опирается только на сам профиль и закон Снеллиуса, поэтому служит независимым критерием физической точности численного решения.
